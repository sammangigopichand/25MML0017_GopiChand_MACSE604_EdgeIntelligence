```python
#open the default camera(0=laptop camera)
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read() # read frame from camera

    if not ret:
        print("Failed to grab frame")
        break

    cv2.imshow("Laptop Camera", frame) # show the frame

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture("samplevideo.mp4")

total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
fps = cap.get(cv2.CAP_PROP_FPS)
width = cap.get(cv2.CAP_PROP_FRAME_WIDTH)
height = cap.get(cv2.CAP_PROP_FRAME_HEIGHT)

print("Total Frames:",total_frames)
print("fps:",fps)
print("Width:",width)
print("height:",height)
```

```python
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    cv2.imshow("Grayscale video",gray)

    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

cap=cv2.VideoCapture("samplevideo1.mp4")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # frame is already pixels
    print("Frame shape:", frame.shape)

    # Example: get pixel at (x=100, y=50)
    pixel = frame[0]
    print("Pixel value (BGR):", pixel)

cap.release()
```

```python
import cv2 

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    edges = cv2.Canny(frame, 100, 200)   # detects the edges

    cv2.imshow("Edges", edges)

    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    negative = 255 - frame

    cv2.imshow("Original", frame)
    cv2.imshow("Negative", negative)

    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
